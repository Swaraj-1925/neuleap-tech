```python
class Customer(Base):
    __tablename__ = 'customers'
    
    customer_id = Column(Integer, primary_key=True)
    full_name = Column(String)
    email = Column(String)
    phone = Column(String)
    country = Column(String)
    signup_date = Column(Date)
    signup_channel = Column(String)
    loyalty_points = Column(Integer)

class Product(Base):
    __tablename__ = 'products'
    
    product_id = Column(Integer, primary_key=True)
    product_name = Column(String)
    price = Column(Numeric)
    category = Column(String)
    brand = Column(String)
    created_at = Column(TIMESTAMP)
    updated_at = Column(TIMESTAMP)

class Order(Base):
    __tablename__ = 'orders'
    
    order_id = Column(Integer, primary_key=True)
    customer_id = Column(Integer, ForeignKey('customers.customer_id')) 
    order_date = Column(Date)
    status = Column(String)
    discount_amount = Column(Numeric)
    tax_amount = Column(Numeric)
    payment_method = Column(String)
    shipping_address = Column(Text)

class OrderItem(Base):
    __tablename__ = 'order_items'
    
    order_item_id = Column(Integer, primary_key=True)
    order_id = Column(Integer, ForeignKey('orders.order_id'))
    product_id = Column(Integer, ForeignKey('products.product_id'))
    quantity = Column(Integer)
    unit_price = Column(Numeric) 

class Payment(Base):
    __tablename__ = 'payments'
    
    payment_id = Column(Integer, primary_key=True)
    order_id = Column(Integer, ForeignKey('orders.order_id'))
    payment_date = Column(Date) 
    payment_amount = Column(Numeric)
    payment_status = Column(String)
    payment_type = Column(String)
```

```python
def load_data():
    directory = "./Dataset"
    
    print("Resetting Database Schema...")
    Base.metadata.drop_all(engine)
    Base.metadata.create_all(engine)
    print("Schema created successfully.")

    
    file_map = {
        'customers': 'customers',
        'products': 'products',
        'orders': 'orders',
        'order_items': 'order_items',
        'payments': 'payment'
    }

    # 2. Track Valid IDs in memory to filter orphans
    valid_ids = {
        'customers': set(),
        'products': set(),
        'orders': set()
    }

    # 3. Load Order (Parents first -> Children last)
    load_order = ['customers', 'products', 'orders', 'order_items', 'payments']
    id_cols = ['customer_id', 'product_id', 'order_id', 'order_item_id', 'payment_id']

    found_files = {}
    for root, dirs, files in os.walk(directory):
        for file in files:
            name_no_ext = os.path.splitext(file)[0].lower()
            found_files[name_no_ext] = os.path.join(root, file)

    for table in load_order:
        file_key = file_map[table]
        
        if file_key in found_files:
            file_path = found_files[file_key]
            print(f"Processing {table} (from {file_key})...")
            
            try:
                if file_path.endswith('.csv'):
                    df = pd.read_csv(file_path)
                elif file_path.endswith(('.xls', '.xlsx')):
                    df = pd.read_excel(file_path)
                
                for col in df.columns:
                    if 'date' in col or 'created_at' in col or 'updated_at' in col:
                        df[col] = pd.to_datetime(df[col], errors='coerce')

                for col in id_cols:
                    if col in df.columns:
                        df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')
                        if col == table[:-1] + '_id':
                             df = df.dropna(subset=[col])

               
                initial_count = len(df)

                if table == 'orders':
                    df = df[df['customer_id'].isin(valid_ids['customers'])]
                
                elif table == 'order_items':
                    # Only keep items where order_id AND product_id exist
                    df = df[df['order_id'].isin(valid_ids['orders'])]
                    df = df[df['product_id'].isin(valid_ids['products'])]

                elif table == 'payments':
                    # Only keep payments where order_id exists
                    df = df[df['order_id'].isin(valid_ids['orders'])]

                dropped_count = initial_count - len(df)
                if dropped_count > 0:
                    print(f"Dropped {dropped_count} orphan rows (IDs missing in parent tables).")

                if table == 'customers':
                    valid_ids['customers'] = set(df['customer_id'].dropna())
                elif table == 'products':
                    valid_ids['products'] = set(df['product_id'].dropna())
                elif table == 'orders':
                    valid_ids['orders'] = set(df['order_id'].dropna())

                df.to_sql(
                    table, 
                    engine, 
                    if_exists='append',
                    index=False, 
                    method='multi', 
                    chunksize=1000
                )
                print(f"Successfully loaded {len(df)} rows into '{table}'.")

            except Exception as e:
                print(f"Error loading {table}: {e}")
        else:
            print(f"File not found for table: {table}")
```

```python
# Retrieve top 5 highest value completed orders
q1 = """
SELECT 
    o.order_id,
    o.customer_id,
    (SUM(oi.quantity * p.price) - o.discount_amount + o.tax_amount) AS order_amount,
    o.order_date
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
WHERE o.status = 'Completed'
GROUP BY o.order_id, o.customer_id, o.order_date, o.discount_amount, o.tax_amount
ORDER BY order_amount DESC
LIMIT 5;
"""
df1 = pd.read_sql(q1, engine)
print(df1)
```

```python
# List customers with more than 2 orders
q2 = """
SELECT 
    c.customer_id,
    c.full_name,
    COUNT(o.order_id) AS total_orders
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.full_name
HAVING COUNT(o.order_id) > 2;
"""
df2 = pd.read_sql(q2, engine)
print(df2)
```

```python
# Get customers who never placed an order using LEFT JOIN
q3 = """
SELECT 
    c.customer_id,
    c.full_name,
    c.signup_date
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
"""
df3 = pd.read_sql(q3, engine)
print(df3)
```

```python
# Show most expensive product purchased by each customer
q4 = """
SELECT DISTINCT ON (o.customer_id)
    o.customer_id, p.product_name, p.price
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
ORDER BY o.customer_id, p.price DESC;
"""
df4 = pd.read_sql(q4, engine)
print(f"len : {len(df4)} \n 1st 5 : \n{df4.head()}")
```

```python
# Q5: Find order with largest number of items
q5 = """
SELECT 
    order_id, 
    SUM(quantity) as total_items
FROM order_items
GROUP BY order_id
ORDER BY total_items DESC
LIMIT 1;
"""
df5 = pd.read_sql(q5, engine)
print(df5)
```

```python
# Q6: Customers who ordered within 30 days of signup
q6 = """
SELECT 
    c.customer_id, 
    c.full_name, 
    c.signup_date, 
    MIN(o.order_date) as first_order_date,
    -- Calculate difference in days for display
    EXTRACT(DAY FROM (MIN(o.order_date)::timestamp - c.signup_date::timestamp)) as days_difference
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.full_name, c.signup_date
-- Filter: Compare the extracted number of days to 30
HAVING EXTRACT(DAY FROM (MIN(o.order_date)::timestamp - c.signup_date::timestamp)) <= 30;
"""

df6 = pd.read_sql(q6, engine)
print(f"len : {len(df6)} \n 1st 5 : \n{df6.head()}")
```

```python
# Q7: Count orders per status per country
q7 = """
SELECT 
    c.country, 
    o.status, 
    COUNT(o.order_id) as order_count
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
GROUP BY c.country, o.status
ORDER BY c.country, order_count DESC;
"""
df7 = pd.read_sql(q7, engine)
print(df7)
```

```python
# Q8: Calculate Customer Lifetime Value (CLV)
q8 = """
SELECT 
    c.customer_id, 
    c.full_name, 
    SUM(oi.quantity * p.price) as total_revenue
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
GROUP BY c.customer_id, c.full_name
ORDER BY total_revenue DESC;
"""
df8 = pd.read_sql(q8, engine)
print(f"len : {len(df8)} \n 1st 5 : \n{df8.head()}")
```

```python
# Q9: Identify repeat customers (>1 order per month)
q9 = """
SELECT 
    c.customer_id, 
    c.full_name, 
    TO_CHAR(o.order_date, 'YYYY-MM') as order_month,
    COUNT(o.order_id) as monthly_order_count
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.full_name, order_month
HAVING COUNT(o.order_id) > 1
ORDER BY monthly_order_count DESC;
"""
df9 = pd.read_sql(q9, engine)
print(df9)
```

```python
q10 = """
SELECT 
    EXTRACT(YEAR FROM o.order_date) as year,
    EXTRACT(MONTH FROM o.order_date) as month,
    SUM(oi.quantity * p.price) as total_revenue
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
GROUP BY 1, 2
ORDER BY 1, 2;
"""
df10 = pd.read_sql(q10, engine)
print(df10)
```

```python
# Q11: Most returned products
q11 = """
SELECT 
    p.product_id, 
    p.product_name, 
    COUNT(*) as return_count
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
JOIN orders o ON oi.order_id = o.order_id
WHERE o.status = 'Returned'
GROUP BY p.product_id, p.product_name
ORDER BY return_count DESC;
"""
df11 = pd.read_sql(q11, engine)
print(df11)
```

```python
q12 = """
SELECT 
    p.category, 
    SUM(oi.quantity * p.price) as total_revenue
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
GROUP BY p.category
ORDER BY total_revenue DESC
LIMIT 3;
"""
df12 = pd.read_sql(q12, engine)
print(df12)
```

```python
# Q13: Basket Analysis (Products frequently bought together)
q13 = """
SELECT 
    a.product_id AS product_id_1, 
    b.product_id AS product_id_2, 
    COUNT(*) AS times_bought_together
FROM order_items a
JOIN order_items b ON a.order_id = b.order_id AND a.product_id < b.product_id
GROUP BY 1, 2
ORDER BY times_bought_together DESC
LIMIT 10;
"""
df13 = pd.read_sql(q13, engine)
print(df13)
```

```python
# Q14: Customer Retention (Ordered again within 90 days)

q14 = """
WITH NextOrders AS (
    SELECT 
        customer_id,
        order_date AS first_order_date,
        LEAD(order_date) OVER (
            PARTITION BY customer_id 
            ORDER BY order_date
        ) AS next_order_date
    FROM orders
)
SELECT 
    c.customer_id, 
    c.full_name, 
    n.first_order_date, 
    n.next_order_date, 
    (n.next_order_date - n.first_order_date) AS days_between_orders
FROM customers c
JOIN NextOrders n 
    ON c.customer_id = n.customer_id
WHERE 
    (n.next_order_date - n.first_order_date) <= 90;

"""

df14 = pd.read_sql(q14, engine)
print(df14)
```

```python
# Q15: Net Revenue After Discounts & Tax monthly country-wise
# Net Revenue = Gross (Item Level) - Discount (Order Level) + Tax (Order Level)
q15 = """
SELECT 
    c.country,
    EXTRACT(YEAR FROM o.order_date) as year,
    EXTRACT(MONTH FROM o.order_date) as month,
    SUM(oi.quantity * p.price) as gross_revenue,
    SUM(o.discount_amount) as total_discount,
    SUM(o.tax_amount) as total_tax,
    (SUM(oi.quantity * p.price) - SUM(o.discount_amount) + SUM(o.tax_amount)) as net_revenue
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
GROUP BY 1, 2, 3
ORDER BY 1, 2, 3;
"""
df15 = pd.read_sql(q15, engine)
print(df15)
```

```python
# Q16: Discount Sensitivity (Customers who ONLY order with discounts)
q16 = """
SELECT 
    c.customer_id, 
    c.full_name, 
    COUNT(o.order_id) as total_orders,
    COUNT(o.order_id) FILTER (WHERE o.discount_amount > 0) as discounted_orders
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.full_name
HAVING COUNT(o.order_id) = COUNT(o.order_id) FILTER (WHERE o.discount_amount > 0);
"""
df16 = pd.read_sql(q16, engine)
print(f"len : {len(df16)} \n 1st 5 : \n{df16.head()}")
```

```python
# Q17: First vs Most Recent Purchase Product (Using ROW_NUMBER)
q17 = """
WITH RankedPurchases AS (
    SELECT 
        o.customer_id, 
        p.product_name,
        ROW_NUMBER() OVER(PARTITION BY o.customer_id ORDER BY o.order_date ASC, oi.order_item_id ASC) as first_rank,
        ROW_NUMBER() OVER(PARTITION BY o.customer_id ORDER BY o.order_date DESC, oi.order_item_id DESC) as last_rank
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
)
SELECT 
    customer_id, 
    MAX(CASE WHEN first_rank = 1 THEN product_name END) as first_product_name,
    MAX(CASE WHEN last_rank = 1 THEN product_name END) as last_product_name
FROM RankedPurchases
GROUP BY customer_id;
"""
df17 = pd.read_sql(q17, engine)
print(f"len : {len(df17)} \n 1st 5 : \n{df17.head()}")
```

```python
# Q18: Pareto Revenue Analysis (80/20 Rule)
q18 = """
WITH CustomerRevenue AS (
    SELECT 
        c.customer_id, c.full_name,
        SUM(oi.quantity * p.price) as customer_revenue
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
    GROUP BY 1, 2
),
RunningTotals AS (
    SELECT *,
        SUM(customer_revenue) OVER (ORDER BY customer_revenue DESC) as cumulative_revenue,
        SUM(customer_revenue) OVER () as total_revenue
    FROM CustomerRevenue
)
SELECT *,
    ROUND(((cumulative_revenue / total_revenue) * 100)::numeric, 2) as cumulative_percentage
FROM RunningTotals
WHERE (cumulative_revenue - customer_revenue) / total_revenue <= 0.80;
"""
df18 = pd.read_sql(q18, engine)
print(f"len : {len(df18)} \n 1st 5 : \n{df18.head()}")
```

```python
# Q19: Brand Affinity per Customer
q19 = """
SELECT 
    o.customer_id, 
    p.brand, 
    SUM(oi.quantity * p.price) as brand_revenue
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
GROUP BY o.customer_id, p.brand
ORDER BY o.customer_id, brand_revenue DESC;
"""
df19 = pd.read_sql(q19, engine)
print(f"len : {len(df19)} \n 1st 5 : \n{df19.head()}")
```

```python
# Q20: Payment Failure Impact (Failed then Successful)
q20 = """
SELECT 
    p1.order_id, 
    o.customer_id,
    COUNT(p1.payment_id) as failed_attempts,
    MAX(p2.payment_date) as successful_payment_date
FROM payment p1
JOIN payment p2 ON p1.order_id = p2.order_id
JOIN orders o ON p1.order_id = o.order_id
WHERE p1.payment_status = 'Failed' 
  AND p2.payment_status = 'Success'
  AND p2.payment_date >= p1.payment_date
GROUP BY p1.order_id, o.customer_id;
"""

df20 = pd.read_sql(q20, engine)
print(df20)
```

```python
# Q21: Customer Order Velocity (Avg days between orders)

# the output is empty beacuse the dataset contains coustomer who has only ordered once
q21 = """
SELECT 
    customer_id,
    full_name,
    COUNT(*) AS total_orders,
    AVG(days_between) AS avg_days_between_orders
FROM (
    SELECT 
        c.customer_id,
        c.full_name,
        o.order_date - LAG(o.order_date) OVER (
            PARTITION BY c.customer_id ORDER BY o.order_date
        ) AS days_between
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
) t
WHERE days_between IS NOT NULL
GROUP BY customer_id, full_name;
"""
df21 = pd.read_sql(q21, engine)
print(df21)
```
